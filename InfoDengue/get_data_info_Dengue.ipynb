```python
import requests
import pandas as pd

# Parâmetros de busca (Ex: Estação de São Paulo A701)
data_inicio = "2026-01-01"
data_fim = "2026-01-15"
codigo_estacao = "A771"

url = f"https://apitempo.inmet.gov.br/estacao/{data_inicio}/{data_fim}/{codigo_estacao}"
print(url)

response = requests.get(url)

if response.status_code == 200:
    dados = response.json()
    # Convertendo para DataFrame para manipular facilmente
    df = pd.DataFrame(dados)
    
    # Selecionando apenas as colunas de interesse
    df_umidade = df[['DT_MEDICAO', 'HR_MEDICAO', 'UMID_REL', 'UMI_MAX', 'UMI_MIN']]
    print(df_umidade.head())
else:

    print(response)

    # print(f"Erro ao acessar API: {response.status_code}")
```
