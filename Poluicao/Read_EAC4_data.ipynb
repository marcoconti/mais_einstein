```python
# DF's usados para calcular a umidade

# Para este calculo será utilizado a formulação empírica de Magnus-Tetens

df_temperatura = \
    (spark.read.csv("C:\\Marco Conti\\Projetos\\MAIS-v2\\dados\\ERA5_temperatura_t2m.csv"
                   ,header=True
                   ,sep=","
                   ,inferSchema=True))

df_ponto_orvalho = \
    (spark.read.csv("C:\Marco Conti\Projetos\MAIS-v2\dados\ERA5_ponto_orvalho.csv"
                   ,header=True
                   ,sep=","
                   ,inferSchema=True))
```

```python
df_temperatura.printSchema()
df_ponto_orvalho.printSchema()
```

```python
# Faz a junção dos dados de tempetarua e ponto de orvalho para calcular a umidade:
print("Temperatura: ", df_temperatura.count())          # 151663
print("Ponto de orvalho:", df_ponto_orvalho.count())    # 151663

df_temp_ponto_orvalho = \
    (df_temperatura.alias('t')
        .join(df_ponto_orvalho.alias('p')
             ,((F.col('t.valid_time') == F.col('p.valid_time')) & 
               (F.col("t.latitude")     == F.col("p.latitude")) & 
               (F.col("t.longitude")    == F.col("p.longitude")))
             ,'inner')
        .select('t.valid_time'
               ,'t.latitude'
               ,'t.longitude'
               ,F.col('t.t2m_celsius').alias('temperatura_celsius')
               ,F.col('p.d2m_celsius').alias('temp_ponto_orvalho_celsius'))
    )

print("Join:", df_temp_ponto_orvalho.count())
```

```python

# Constantes empíricas da equação de Magnus-Tetens.
A = 17.67
B = 243.5

df_umidade_Magnus_Tetens = (
    df_temp_ponto_orvalho
    .withColumn(
        "umidade_relativa",
        F.round(  
            F.lit(100.0) * F.exp(
                (
                    A * F.col("temp_ponto_orvalho_celsius") /
                    (F.col("temp_ponto_orvalho_celsius") + B)
                )
                -
                (
                    A * F.col("temperatura_celsius") /
                    (F.col("temperatura_celsius") + B)
                )
            )
        ,4)
    )
)

df_umidade_Magnus_Tetens.show()
```

```python

def vapor_pressure_buck(col_temp):
    return (
        6.1121 *
        F.exp(
            ((18.678 - (col_temp / 234.5)) * col_temp)
            /
            (257.14 + col_temp)
        )
    )

df_umidade_Arden_Buck = (
    df_temp_ponto_orvalho
    .withColumn("e_td", vapor_pressure_buck(F.col("temp_ponto_orvalho_celsius")))
    .withColumn("e_t", vapor_pressure_buck(F.col("temperatura_celsius")))
    .withColumn(
        "umidade_relativa",
        F.round(
            100 * F.col("e_td") / F.col("e_t"),
            4
        )
    )
    .drop("e_td", "e_t")
)

df_umidade_Arden_Buck.show()

```
