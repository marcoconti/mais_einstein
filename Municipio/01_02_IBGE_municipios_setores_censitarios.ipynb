```python
setores_censitarios = geobr.read_census_tract(
    year=2022
)
```

```python
# converte geometria para texto WKT
setores_censitarios["geometry_wkt"] = setores_censitarios.geometry.to_wkt()

# remove coluna geometry do geopandas
pdf = setores_censitarios.drop(columns=["geometry"])

# cria Spark DataFrame
df_spark = spark.createDataFrame(pdf)
```
